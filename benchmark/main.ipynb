```python
import numpy as np
import matplotlib.pyplot as plt
import d3rlpy

# ============================================================
# INPUTS
# ============================================================

IQL_PATH = r"iql_runs/iql_tau0.9_wt5/checkpoints/epoch_0020.d3"
CALQL_PATH = r"calql_runs/calql_cw1_gamma0.95/checkpoints/epoch_0075.d3"

DEVICE = "cuda:0"

# Load models
iql = d3rlpy.load_learnable(IQL_PATH, device=DEVICE)
calql = d3rlpy.load_learnable(CALQL_PATH, device=DEVICE)


# ============================================================
# GET STATES FROM AN IQL ROLLOUT
# ============================================================

def collect_rollout_states(env, agent, seed=0):
    """
    Runs one trajectory using 'agent' and returns all observations
    encountered BEFORE each action.
    """
    
    obs, info = env.reset(seed=seed)

    states = []
    actions = []

    terminated = False
    truncated = False

    while not (terminated or truncated):

        states.append(np.asarray(obs, dtype=np.float32).copy())

        action = agent.predict(
            np.asarray(obs, dtype=np.float32)[None, :]
        )[0]

        actions.append(action.copy())

        obs, reward, terminated, truncated, info = env.step(action)

    return np.asarray(states), np.asarray(actions)


# ============================================================
# CREATE ENVIRONMENT
# ============================================================

# Change import if needed
from poly_reactor_env import PolymerisationEnv

env = PolymerisationEnv()


# Use IQL to generate the reference state trajectory
states, iql_rollout_actions = collect_rollout_states(
    env,
    iql,
    seed=0,
)


# ============================================================
# ASK BOTH POLICIES WHAT THEY WOULD DO
# AT EXACTLY THE SAME STATES
# ============================================================

iql_actions = iql.predict(states)
calql_actions = calql.predict(states)


# ============================================================
# PLOT
# ============================================================

steps = np.arange(len(states))

fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)

# Initiator
axes[0].plot(
    steps,
    iql_actions[:, 0],
    label="IQL"
)

axes[0].plot(
    steps,
    calql_actions[:, 0],
    label="CAL-QL"
)

axes[0].set_ylabel("u_I")
axes[0].set_title("Initiator action on identical states")
axes[0].legend()
axes[0].grid(alpha=0.3)


# Coolant
axes[1].plot(
    steps,
    iql_actions[:, 1],
    label="IQL"
)

axes[1].plot(
    steps,
    calql_actions[:, 1],
    label="CAL-QL"
)

axes[1].set_ylabel("u_Tc")
axes[1].set_xlabel("Environment step")
axes[1].set_title("Coolant action on identical states")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# NUMERICAL COMPARISON
# Look specifically at later steady-state region
# ============================================================

late_start = int(0.4 * len(states))

print("Late trajectory mean actions")
print("--------------------------------")

print(
    "IQL:",
    np.mean(iql_actions[late_start:], axis=0)
)

print(
    "CAL-QL:",
    np.mean(calql_actions[late_start:], axis=0)
)

print("\nMean absolute action difference")
print("--------------------------------")

print(
    np.mean(
        np.abs(iql_actions - calql_actions),
        axis=0
    )
)
```

```python
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# 1. PICK STEADY-STATE OBSERVATION
# ============================================================

state_idx = 100

state = states[state_idx].copy()

print("Selected state:")
print(state)


# ============================================================
# 2. ACTION SWEEP
# ============================================================

# Sweep initiator action
u_I_values = np.linspace(0.60, 0.90, 301)

# Fix coolant at IQL's action for this exact state
u_Tc_fixed = iql_actions[state_idx, 1]

actions = np.column_stack([
    u_I_values,
    np.full_like(u_I_values, u_Tc_fixed)
]).astype(np.float32)

# Repeat identical state for every candidate action
state_batch = np.repeat(
    state[None, :],
    len(actions),
    axis=0
).astype(np.float32)


# ============================================================
# 3. CAL-QL Q VALUES
# ============================================================

q_values = calql.predict_value(
    state_batch,
    actions
)

q_values = np.asarray(q_values).reshape(-1)


# ============================================================
# 4. ACTOR OUTPUTS
# ============================================================

calql_action = calql.predict(
    state[None, :]
)[0]

iql_action = iql.predict(
    state[None, :]
)[0]


# ============================================================
# 5. FIND CRITIC'S PREFERRED ACTION
# ============================================================

best_idx = np.argmax(q_values)
best_u_I = u_I_values[best_idx]


# ============================================================
# 6. PLOT
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    u_I_values,
    q_values,
    label="CAL-QL critic"
)

plt.axvline(
    best_u_I,
    linestyle="--",
    label=f"Critic maximum = {best_u_I:.3f}"
)

plt.axvline(
    calql_action[0],
    linestyle=":",
    label=f"CAL-QL actor = {calql_action[0]:.3f}"
)

plt.axvline(
    iql_action[0],
    linestyle="-.",
    label=f"IQL actor = {iql_action[0]:.3f}"
)

plt.xlabel("u_I")
plt.ylabel("Q(s,a)")

plt.title(
    f"CAL-QL critic action landscape\n"
    f"state = IQL trajectory step {state_idx}, "
    f"u_Tc = {u_Tc_fixed:.3f}"
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# 7. NUMERICAL RESULTS
# ============================================================

print("\n--- Results ---")
print(f"IQL actor u_I:       {iql_action[0]:.4f}")
print(f"CAL-QL actor u_I:    {calql_action[0]:.4f}")
print(f"Critic best u_I:     {best_u_I:.4f}")

# Also compare Q at the two actual actor actions
iql_test_action = np.array(
    [[iql_action[0], u_Tc_fixed]],
    dtype=np.float32
)

calql_test_action = np.array(
    [[calql_action[0], u_Tc_fixed]],
    dtype=np.float32
)

single_state = state[None, :].astype(np.float32)

q_iql_action = calql.predict_value(
    single_state,
    iql_test_action
)[0]

q_calql_action = calql.predict_value(
    single_state,
    calql_test_action
)[0]

print(f"\nQ at IQL u_I:        {q_iql_action:.6f}")
print(f"Q at CAL-QL u_I:     {q_calql_action:.6f}")
print(
    f"Q difference:        "
    f"{q_calql_action - q_iql_action:.6f}"
)
```

```python
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# SETTINGS
# ============================================================

# Pick a state well inside the steady-state region
state_idx = 100

# Initiator action range to investigate
u_I_values = np.linspace(0.60, 0.90, 301)


# ============================================================
# GET EXACT SAME STATE FOR BOTH CRITICS
# ============================================================

state = states[state_idx].copy()

# Fix coolant action at IQL's action at this state
u_Tc_fixed = iql_actions[state_idx, 1]

print("Selected state index:", state_idx)
print("Fixed u_Tc:", u_Tc_fixed)


# ============================================================
# BUILD ACTION SWEEP
# ============================================================

actions = np.column_stack([
    u_I_values,
    np.full_like(u_I_values, u_Tc_fixed)
]).astype(np.float32)

# Same state repeated for every candidate action
state_batch = np.repeat(
    state[None, :],
    len(actions),
    axis=0
).astype(np.float32)


# ============================================================
# GET Q VALUES FROM BOTH CRITICS
# ============================================================

q_iql = iql.predict_value(
    state_batch,
    actions
)

q_calql = calql.predict_value(
    state_batch,
    actions
)

q_iql = np.asarray(q_iql).reshape(-1)
q_calql = np.asarray(q_calql).reshape(-1)


# ============================================================
# FIND CRITIC OPTIMA
# ============================================================

iql_best_idx = np.argmax(q_iql)
calql_best_idx = np.argmax(q_calql)

iql_best_u_I = u_I_values[iql_best_idx]
calql_best_u_I = u_I_values[calql_best_idx]


# Actual actor outputs at EXACT SAME STATE
iql_actor_action = iql.predict(
    state[None, :]
)[0]

calql_actor_action = calql.predict(
    state[None, :]
)[0]


# ============================================================
# PLOT 1: IQL CRITIC
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    u_I_values,
    q_iql,
    label="IQL critic"
)

plt.axvline(
    iql_best_u_I,
    linestyle="--",
    label=f"IQL critic maximum = {iql_best_u_I:.3f}"
)

plt.axvline(
    iql_actor_action[0],
    linestyle=":",
    label=f"IQL actor = {iql_actor_action[0]:.3f}"
)

plt.axvline(
    calql_actor_action[0],
    linestyle="-.",
    label=f"CAL-QL actor = {calql_actor_action[0]:.3f}"
)

plt.xlabel("u_I")
plt.ylabel("Q(s,a)")

plt.title(
    f"IQL critic action landscape\n"
    f"state = IQL trajectory step {state_idx}, "
    f"u_Tc = {u_Tc_fixed:.3f}"
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# PLOT 2: CAL-QL CRITIC
# ============================================================

plt.figure(figsize=(9, 5))

plt.plot(
    u_I_values,
    q_calql,
    label="CAL-QL critic"
)

plt.axvline(
    calql_best_u_I,
    linestyle="--",
    label=f"CAL-QL critic maximum = {calql_best_u_I:.3f}"
)

plt.axvline(
    calql_actor_action[0],
    linestyle=":",
    label=f"CAL-QL actor = {calql_actor_action[0]:.3f}"
)

plt.axvline(
    iql_actor_action[0],
    linestyle="-.",
    label=f"IQL actor = {iql_actor_action[0]:.3f}"
)

plt.xlabel("u_I")
plt.ylabel("Q(s,a)")

plt.title(
    f"CAL-QL critic action landscape\n"
    f"state = IQL trajectory step {state_idx}, "
    f"u_Tc = {u_Tc_fixed:.3f}"
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# NUMERICAL SUMMARY
# ============================================================

print("\n==============================")
print("CRITIC COMPARISON")
print("==============================")

print(f"IQL actor u_I:          {iql_actor_action[0]:.4f}")
print(f"CAL-QL actor u_I:       {calql_actor_action[0]:.4f}")

print()

print(f"IQL critic best u_I:    {iql_best_u_I:.4f}")
print(f"CAL-QL critic best u_I: {calql_best_u_I:.4f}")


# ============================================================
# Q VALUES AT THE TWO ACTOR ACTIONS
# ============================================================

iql_actor_test = np.array(
    [[iql_actor_action[0], u_Tc_fixed]],
    dtype=np.float32
)

calql_actor_test = np.array(
    [[calql_actor_action[0], u_Tc_fixed]],
    dtype=np.float32
)

single_state = state[None, :].astype(np.float32)


# What does IQL critic think about both actions?
iql_Q_at_iql = iql.predict_value(
    single_state,
    iql_actor_test
)[0]

iql_Q_at_calql = iql.predict_value(
    single_state,
    calql_actor_test
)[0]


# What does CAL-QL critic think about both actions?
calql_Q_at_iql = calql.predict_value(
    single_state,
    iql_actor_test
)[0]

calql_Q_at_calql = calql.predict_value(
    single_state,
    calql_actor_test
)[0]


print("\n==============================")
print("IQL CRITIC")
print("==============================")

print(f"Q(IQL action):      {iql_Q_at_iql:.6f}")
print(f"Q(CAL-QL action):   {iql_Q_at_calql:.6f}")

print(
    "Preference:",
    "IQL action"
    if iql_Q_at_iql > iql_Q_at_calql
    else "CAL-QL action"
)


print("\n==============================")
print("CAL-QL CRITIC")
print("==============================")

print(f"Q(IQL action):      {calql_Q_at_iql:.6f}")
print(f"Q(CAL-QL action):   {calql_Q_at_calql:.6f}")

print(
    "Preference:",
    "IQL action"
    if calql_Q_at_iql > calql_Q_at_calql
    else "CAL-QL action"
)
```
